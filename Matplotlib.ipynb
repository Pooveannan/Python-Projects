```python
import pandas as pd

df = pd.read_excel("Data Visualization Dataset.xlsx", engine="openpyxl")
df.head()  # Preview the first few rows
```

```python
df["date"] = pd.to_datetime(df["date"])
df = df.sort_values(by="date")  # Sort by date for proper visualization
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
plt.plot(df["date"], df["sales_amount"], marker='o', linestyle='-', label="sales_amount", color='blue')
plt.plot(df["date"], df["marketing_spend"], marker='s', linestyle='-', label="marketing_spend", color='green')

plt.xlabel("date")
plt.ylabel("amount")
plt.title("Sales Amount & Marketing Spend Over Time")
plt.legend()
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt

df_grouped = df.groupby("product")["sales_amount"].sum().reset_index()

# Sort values for better visualization
df_grouped = df_grouped.sort_values(by="sales_amount", ascending=True)  # Ascending for horizontal bars

df_grouped.head()
plt.figure(figsize=(10, 6))
plt.barh(df_grouped["product"], df_grouped["sales_amount"], color="skyblue")

plt.xlabel("Sum of Sales Amount")
plt.ylabel("Product")
plt.title("Top-Selling & Low-Performing Products")
plt.grid(axis="x", linestyle="--", alpha=0.7)

plt.show()
```
