```python

## list in python

my_list = [1,2,3,4] #list
print(my_list)
my_list = my_list*2 #product of list
print(my_list)


## numpy array

import numpy as np
array = np.array([1,2,3,4,5]) #creating array using numpy
print(array)
print(type(array))

#product in numpy array
array = array * 2
print(array)
```

```python
## Multi-dimensional Arrays

import numpy as np
array = np.array('A') # giving some value A in the array, this is a 0-Dimensional array
print(array.ndim) # .ndim gives the number of dimensions in an array

array2 = np.array(['A','B','C']) # array2 is a 1-D array
print(array2.ndim)

array3 = np.array([['A','B','C'],
                   ['d','e','f'],
                   ['z','y','x']])
print(array3.ndim)

array4 = np.array([[['A','B','C'], ['d','e','f'], ['z','y','x']],
                   [['A','B','C'], ['d','e','f'], ['z','y','x']],
                   [['A','B','C'], ['d','e','f'], ['z','y','x']]]) # 3-D array

print(array4.ndim)
```
