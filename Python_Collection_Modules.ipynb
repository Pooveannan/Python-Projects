```python
#For checking permissions

def permission_required(func):
 def wrapper(is_allowed):
  if is_allowed:
   func()
  else:
   print ("Permission denied")
 return wrapper
```

```python
# To measure time

import time
def time_taken(func):
 def wrapper():
  start = time.time()
  func()
  end = time.time()
  print("Time taken: ", end - start)
 return wrapper
```

```python
@time_taken
def study():
 print("Studying python decorators")
time.sleep(2)
```

```python
## Decorators with Arguments

def repeat_dec(times):
 def actual_dec(func):
  def wrapper():
   for i in range(times):
    print(f"Execution number{i+1}: ")
    func()
  return wrapper
 return actual_dec
```

```python
@repeat_dec(5)
def greet():
 print("Hello World!")
```

```python
greet()
```

```python
### Application of Decorators with Arguments:-

## Example 1: Role-Based Permissions;

def permission_required(role):
 def decorator(func):
  def wrapper(user_role):
   if user_role == role:
    func()
   else:
    print("Access Denied!!!")
  return wrapper
 return decorator
```

```python
@permission_required("Admin")
def show_dashboard():
 print("Showing dashboard...")
```

```python
show_dashboard("Admin")
```

```python
show_dashboard("Student")
```

```python
### Application of Decorators with Arguments:-

## Example 2: Limit number of calls:

def limit_calls(max_calls):
 def decorator(func):
  count = 0
  def wrapper():
   nonlocal count
   if count < max_calls:
    count = count + 1
    func()
   else:
    print("Limit reached!!!")
  return wrapper
 return decorator
```

```python
@limit_calls(2)
def send_messages():
 print("Message sent.")
```

```python
send_messages()
```

```python
send_messages()
```

```python
send_messages()
```

```python

### Application of Decorators with Arguments
## Example 3: Minimum Marks Validation

def min_mark_required(min_marks):
 def decorator(func):
  def wrapper(marks):
   if marks >= min_marks:
    func(marks)
   else:
    print("Failed!")
  return wrapper
 return decorator
```

```python
@min_mark_required(40)
def show_result(marks):
 print(f"Passed with {marks} marks")
```

```python
show_result(88)
```

```python
show_result(26)
```

```python
### Collection Modules

## 1) Counter:-

from collections import Counter
fruits = ["apple", "apple","apple","orange","orange","banana","mango"]
count = Counter(fruits)
print(count)
```

```python
## Example for Counter function:-

text = "Datascience"
count = Counter(text)
count
```

```python
## Example for .most_common()

marks = [90,80,90,75,90,70,80,69,50,50]
count = Counter(marks)
print(count.most_common(3))
```

```python
##Example for .elements()

fruits = ["apple", "apple","apple","orange","orange","banana","mango"]
count = Counter(fruits)
print(list(count.elements()))
```

```python
## Example for .update()

lang = ["python", "Java", "python"]
count = Counter(lang)
print(count)
count.update(["python","java","c++"])
print(count)
```

```python
##Example for .subtract()

fruits = ["apple", "apple","apple","orange","orange","banana","mango"]
count = Counter(fruits)
print(count)
count.subtract({"apple":2, "mango":2})
print(count)
```

```python
 ## Example for Addition (+)

C1 = Counter(a=5, b=8)
C2 = Counter(a=2, b=10)
print(C1+C2)
print(C1-C2) ##negative integers wont be coming in output, b= -2
```

```python
## Example of Intersection:

from collections import Counter

c1 = Counter(a=4,b=2,c=1)
c2 = Counter(a=2,b=3)
print(c1&c2)
```

```python
## Example of Union:

from collections import Counter

c1 = Counter(a=4,b=2,c=1)
c2 = Counter(a=2,b=3)
print(c1|c2)
```

```python
### Collection Modules - 2) namedtuple()

## Example for namedtuple-

from collections import namedtuple
Person= namedtuple("Person", ['name','age','city'])
p1 = Person("Rahul",27,"Bangalore")
print(p1.name)
print(p1.age)
print(p1.city)
```

```python
## 2nd Example for namedtuple: Student Marks

Student = namedtuple("Student",['id','name','marks'])
s1 = Student(101,"Reze",78)
s2 = Student(102,"Shanks",95)
s3 = Student(103,"Alex",90)
print(s1)
print(s1.marks)
```

```python
### Useful attributes/methods of namedtuple-

## 1) Example for _fields

Point = namedtuple("Point", ['x','y'])
p = Point(10,20)
print(p._fields)
```

```python
### Useful attributes/methods of namedtuple-

## 2) Example for _replace

p1 = Point(12,24)
p2 = p1._replace(y=30)
print(p1)
print(p2)
```

```python
### Useful attributes/methods of namedtuple-

## 3) Example for _asdict

Product= namedtuple("Product",['item','price'])
prod1= Product("pen",15)
print(prod1)
print(prod1._asdict())
```

```python
### Useful attributes/methods of namedtuple-

## 4) Example for _make

Point = namedtuple("Point", ['x','y'])
values = [10,30]
p = Point._make(values)
print(p)
```

```python
### Useful attributes/methods of namedtuple-

## 4) Example2 for _make

Employee = namedtuple("Employee",['id','name','salary'])
emp_detail = (1001, "George", 25000)
emp = Employee._make(emp_detail)
print(emp)
```

```python
### Collection Modules

### 3) defaultdict-
## counting items example(int):-

from collections import defaultdict
fruits = ["apple", "apple","apple","orange","orange","banana","mango"]
count = defaultdict(int)
print(count)
for fruit in fruits:
 count[fruit] += 1
print(count)
```

```python
### Collection Modules

### 3) defaultdict
## Grouping Values example(list):-

students = [('Rahul','CSE'), ("Alex","ECE"), ('George','CSE'),("Maya","ECE")]
dept_students = defaultdict(list)
for name,dept in students:
 dept_students[dept].append(name)
print(dept_students)
```

```python
### Collection Modules

### 3) defaultdict
## Grouping Values example(set):-

courses = [('Rahul','Python'), ("Alex","SQL"), ("Rahul","Python"),("Maya","ML")]
student_course = defaultdict(set)
for name,course in courses:
 student_course[name].add(course)
print(student_course)
```
